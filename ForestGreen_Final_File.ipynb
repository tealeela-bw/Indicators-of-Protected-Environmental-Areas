</p>

<p style="margin-bottom: 0;">
The aim of this project is to recognise globally generalisable factors that potentially influence the frequency at which nations across the world prioritise protected areas within their borders. Identifying possible risk-factors could contribute to the development of more comprehensive frameworks for protective areas, potentially assisting the UN’s 13th Sustainable Development Goal (SDG); Climate Action. By using large datasets from the World Bank we will scan for patterns and variables that predict outcomes related to protective areas, contributing to the development of protective schemes likely to yield environmental, economic, and psychological benefits.

```python
import numpy as np

# Load dataset
df = pd.read_csv('combined_wdi_countries_only.csv')

# Define custom variable labels
variable_descriptions = {
    "GOV_WGI_PV_SC": "Political Stability (SCORE 0-100)",
    "GOV_WGI_CC_EST": "Control of Corruption (governance estimate)",
    "SL_TLF_TOTL_FE_ZS": "Female Share of the Labour Force (%)",
    "SG_GEN_PARL_ZS": "Women's Share of Parliamentary Seats (%)",
    "ER_PTD_TOTL_ZS": "Protected Areas (% of total land and marine area)",
    "NY_GDP_MINR_RT_ZS": "Mineral Rents (% of GDP)",
    "NY_GDP_TOTL_RT_ZS": "Total Natural Resources Rents (% OF GDP)",
}

# ==========================================
# 1. MISSING VALUES ASSESSMENT
# ==========================================
missing_report = []

for code, label in variable_descriptions.items():
    if code in df.columns:
        missing_count = df[code].isnull().sum()
        missing_pct = (missing_count / len(df)) * 100
        missing_report.append({
            "Variable Code": code,
            "Indicator Description": label,
            "Missing Count": missing_count,
            "Missing (%)": round(missing_pct, 2)
        })

df_missing = pd.DataFrame(missing_report)
print("--- 1. MISSING VALUES ---")
print(df_missing.to_string(index=False))


# ==========================================
# 2. INCONSISTENCIES & BOUND CHECKS
# ==========================================
print("\n--- 2. INCONSISTENCIES ---")

# Check A: Primary key duplicates (Country + Year)
duplicate_keys = df.duplicated(subset=['country_code', 'year'], keep=False).sum()
print(f"Duplicate (Country, Year) records: {duplicate_keys}")

# Check B: Out-of-bounds percentage checks (0% to 100%)
percentage_vars = [
    "SL_TLF_TOTL_FE_ZS", 
    "SG_GEN_PARL_ZS", 
    "ER_PTD_TOTL_ZS"
]

for code in percentage_vars:
    if code in df.columns:
        label = variable_descriptions[code]
        invalid_rows = df[(df[code] < 0) | (df[code] > 100)]
        print(f"Out of range (0–100%) in '{label}': {len(invalid_rows)} rows")

# Check C: Negative values in rent indicators (% of GDP cannot be < 0)
rent_vars = ["NY_GDP_MINR_RT_ZS", "NY_GDP_TOTL_RT_ZS"]
for code in rent_vars:
    if code in df.columns:
        label = variable_descriptions[code]
        negative_rows = df[df[code] < 0]
        print(f"Negative values found in '{label}': {len(negative_rows)} rows")


# ==========================================
# 3. OTHER DATA QUALITY ISSUES (OUTLIERS & RANGES)
# ==========================================
print("\n--- 3. OUTLIERS AND RANGE SUMMARY ---")

outlier_report = []

for code, label in variable_descriptions.items():
    if code in df.columns:
        Q1 = df[code].quantile(0.25)
        Q3 = df[code].quantile(0.75)
        IQR = Q3 - Q1
        
        # Outlier condition using 1.5 * IQR rule
        outliers_count = df[(df[code] < (Q1 - 1.5 * IQR)) | (df[code] > (Q3 + 1.5 * IQR))][code].count()
        
        outlier_report.append({
            "Indicator Description": label,
            "Min Value": round(df[code].min(), 2),
            "Max Value": round(df[code].max(), 2),
            "Outliers Count (IQR)": outliers_count
        })

df_outliers = pd.DataFrame(outlier_report)
print(df_outliers.to_string(index=False))
```

```python
from IPython.display import HTML, display
import pandas as pd

# 1. Load dataset and filter panel (2013-2023)
df = pd.read_csv("combined_wdi.csv")
df_panel= df
# 2. Define model variables
dep_var = "ER_PTD_TOTL_ZS"
indep_vars = [
    "SL_TLF_TOTL_FE_ZS",
    "SG_GEN_PARL_ZS",
    "GOV_WGI_PV_SC",
    "GOV_WGI_CC_EST",
    "NY_GDP_MINR_RT_ZS",
    "NY_GDP_TOTL_RT_ZS",
]

all_vars = ["country_code", "country", "year", dep_var] + indep_vars
model_data = df_panel[all_vars].dropna()

# 3. Metadata dictionaries
variable_names = {
    "country_code": "Country Code (ISO-3)",
    "country": "Country Name",
    "year": "Observation Year",
    "ER_PTD_TOTL_ZS": (
        "Terrestrial and Marine Protected Areas (% of total land area)"
    ),
    "SL_TLF_TOTL_FE_ZS": "Female Labor Force Share (% of total labor force)",
    "SG_GEN_PARL_ZS": (
        "Women in National Parliaments (% of seats held by women)"
    ),
    "GOV_WGI_PV_SC": (
        "Political Stability and Absence of Violence/Terrorism (Percentile)"
    ),
    "GOV_WGI_CC_EST": "Control of Corruption (WGI Standardized Estimate)",
    "NY_GDP_MINR_RT_ZS": "Mineral Rents (% of GDP)",
    "NY_GDP_TOTL_RT_ZS": "Total Natural Resource Rents (% of GDP)",
}

variable_roles = {
    "country_code": "Identifier",
    "country": "Identifier",
    "year": "Temporal Index",
    "ER_PTD_TOTL_ZS": "Dependent Variable (DV)",
    "SL_TLF_TOTL_FE_ZS": "Independent Variable (IV)",
    "SG_GEN_PARL_ZS": "Independent Variable (IV)",
    "GOV_WGI_PV_SC": "Independent Variable (IV)",
    "GOV_WGI_CC_EST": "Independent Variable (IV)",
    "NY_GDP_MINR_RT_ZS": "Independent Variable (IV)",
    "NY_GDP_TOTL_RT_ZS": "Independent Variable (IV)",
}

# 4. Construct Data Dictionary rows
data_dict = []
for var in all_vars:
  col_data = model_data[var]
  if pd.api.types.is_numeric_dtype(col_data) and var != "year":
    mean_val = f"{col_data.mean():.2f}"
    std_val = f"{col_data.std():.2f}"
    min_val = f"{col_data.min():.2f}"
    max_val = f"{col_data.max():.2f}"
  elif var == "year":
    mean_val = "N/A"
    std_val = "N/A"
    min_val = str(int(col_data.min()))
    max_val = str(int(col_data.max()))
  else:
    mean_val = "N/A"
    std_val = "N/A"
    min_val = f"{col_data.nunique()} Unique"
    max_val = "N/A"

  data_dict.append({
      "Code": var,
      "Description": variable_names.get(var, var),
      "Role": variable_roles.get(var, "Control"),
      "Count": f"{len(col_data):,}",
      "Mean": mean_val,
      "Std Dev": std_val,
      "Min": min_val,
      "Max": max_val,
  })

dict_df = pd.DataFrame(data_dict)

# 5. Display as styled HTML Table in Jupyter
html_table = (
    dict_df.style.set_properties(**{"padding": "8px", "text-align": "right"})
    .set_properties(
        subset=["Code", "Description", "Role"], **{"text-align": "left"}
    )
    .set_table_styles([
        {
            "selector": "th",
            "props": [
                ("background-color", "#2B4C7E"),
                ("color", "white"),
                ("font-weight", "bold"),
                ("text-align", "center"),
            ],
        },
        {
            "selector": "tr:nth-of-type(even)",
            "props": [("background-color", "#F4F7FA")],
        },
        {
            "selector": "table",
            "props": [("border", "1px solid #E0E0E0"), ("width", "100%")],
        },
    ])
    .hide(axis="index")
    .to_html()
)

display(HTML(f"<h3>Master Dataset Data Dictionary</h3>{html_table}"))
```

Looking at our empirical setup, we decided that an Ordinary Least Squares (OLS) regression was the most effective choice because it provides clear, directly interpretable coefficients (b). It allows us to see clearly the linear relationship between the dependent variable and its independent variable. We can see if the relationship is positive or negative, and how much each data point affects the prediction of territorial protection. This allows us to quantify the exact relationship between our indicators and protected area coverage. 

For missing values, we applied a complete-case analysis. First, we isolated individual countries by removing regional and income aggregates. Next, we restricted our dataset to the 2013–2023 study window where coverage is most complete. Out of 2,387 total country-year observations in the panel, 1,524 complete observations were retained to estimate the final model.

```python
# 1. Load dataset and filter for the 2013-2023 study window
df = pd.read_csv("combined_wdi_countries_only.csv")
df_panel = df[df["year"].between(2013, 2023)].copy()

# 2. Variable mapping dictionary
variable_descriptions = {
    "ER_PTD_TOTL_ZS": "Protected Areas (% total land/marine)",
    "GOV_WGI_PV_SC": "Political Stability (Score 0-100)",
    "GOV_WGI_CC_EST": "Control of Corruption (Est.)",
    "SL_TLF_TOTL_FE_ZS": "Female Labor Force Share (%)",
    "SG_GEN_PARL_ZS": "Women in Parliament (%)",
    "NY_GDP_MINR_RT_ZS": "Mineral Rents (% GDP)",
    "NY_GDP_TOTL_RT_ZS": "Natural Resource Rents (% GDP)",
}

# 3. Define Dependent Variable (DV) and Independent Variables (IVs)
dep_var = "ER_PTD_TOTL_ZS"
indep_vars = [
    "SL_TLF_TOTL_FE_ZS",  # Female Labor Force Share (%)
    "SG_GEN_PARL_ZS",  # Women in Parliament (%)
    "GOV_WGI_PV_SC",  # Political Stability Score (0-100)
    "GOV_WGI_CC_EST",  # Control of Corruption Estimate (-2.5 to 2.5)
    "NY_GDP_MINR_RT_ZS",  # Mineral Rents (% GDP)
    "NY_GDP_TOTL_RT_ZS",  # Natural Resource Rents (% GDP)
]

# 4. Drop rows with missing values across all selected model variables
model_data = df_panel[[dep_var] + indep_vars].dropna()

# 5. Add constant intercept and fit OLS model with robust HC3 standard errors
X = sm.add_constant(model_data[indep_vars])
y = model_data[dep_var]

model = sm.OLS(y, X).fit(cov_type="HC3")

# ==============================================================================
# TABLE 1: MODEL FIT STATISTICS
# ==============================================================================
model_fit_table = pd.DataFrame(
    {
        "Metric / Statistic": [
            "Number of Observations (N)",
            "R-squared (R²)",
            "Adjusted R-squared (Adj. R²)",
            "F-statistic",
            "Prob > F (Model p-value)",
            "Log-Likelihood",
            "Akaike Information Criterion (AIC)",
            "Bayesian Information Criterion (BIC)",
        ],
        "Value": [
            f"{int(model.nobs):,}",
            f"{model.rsquared:.4f}",
            f"{model.rsquared_adj:.4f}",
            f"{model.fvalue:.2f}",
            f"{model.f_pvalue:.4e}",
            f"{model.llf:.2f}",
            f"{model.aic:.2f}",
            f"{model.bic:.2f}",
        ],
    }
)

# ==============================================================================
# TABLE 2: REGRESSION COEFFICIENTS & HYPOTHESIS TESTING
# ==============================================================================
coef_labels = ["Constant (Intercept)"] + [
    variable_descriptions[var] for var in indep_vars
]

coefficients_table = pd.DataFrame(
    {
        "Variable Description": coef_labels,
        "Indicator Code": X.columns,
        "Coefficient (β)": model.params.values,
        "Std. Error": model.bse.values,
        "z-statistic": model.tvalues.values,
        "p-value": model.pvalues.values,
        "95% CI Lower": model.conf_int()[0].values,
        "95% CI Upper": model.conf_int()[1].values,
    }
)

# Format numeric outputs for clean print
coefficients_table["Coefficient (β)"] = coefficients_table[
    "Coefficient (β)"
].map("{:.4f}".format)
coefficients_table["Std. Error"] = coefficients_table["Std. Error"].map(
    "{:.4f}".format
)
coefficients_table["z-statistic"] = coefficients_table["z-statistic"].map(
    "{:.3f}".format
)
coefficients_table["p-value"] = coefficients_table["p-value"].apply(
    lambda p: "< 0.001" if p < 0.001 else f"{p:.4f}"
)
coefficients_table["95% CI"] = coefficients_table.apply(
    lambda row: f"[{row['95% CI Lower']:.3f}, {row['95% CI Upper']:.3f}]", axis=1
)

final_coef_table = coefficients_table[
    [
        "Variable Description",
        "Indicator Code",
        "Coefficient (β)",
        "Std. Error",
        "z-statistic",
        "p-value",
        "95% CI",
    ]
]

# Print out results
#For presentation purposes
from IPython.display import display

# Display formatted HTML tables directly in Jupyter
display(model_fit_table)
display(final_coef_table)
```

```python
# 1. Load dataset and filter panel (2013-2023)
df = pd.read_csv("combined_wdi_countries_only.csv")
df_panel = df[df["year"].between(2013, 2023)].copy()

# 2. Define dependent and independent variables
dep_var = "ER_PTD_TOTL_ZS"
indep_vars = [
    "SL_TLF_TOTL_FE_ZS",  # Female Labor Force Share (%)
    "SG_GEN_PARL_ZS",  # Women in Parliament (%)
    "GOV_WGI_PV_SC",  # Political Stability Score
    "GOV_WGI_CC_EST",  # Control of Corruption Estimate
    "NY_GDP_MINR_RT_ZS",  # Mineral Rents (% GDP)
    "NY_GDP_TOTL_RT_ZS",  # Natural Resource Rents (% GDP)
]

# Drop missing values across model variables
model_data = df_panel[[dep_var] + indep_vars].dropna()

# 3. Fit Master OLS Model (HC3 Robust)
X = sm.add_constant(model_data[indep_vars])
y = model_data[dep_var]
model = sm.OLS(y, X).fit(cov_type="HC3")

# 4. Filter for ONLY statistically significant variables (p < 0.001)
sig_vars = [
    (
        "SL_TLF_TOTL_FE_ZS",
        "Female Labor Force Share (%)",
        "#8E44AD",
    ),  # Purple
    ("SG_GEN_PARL_ZS", "Women in Parliament (%)", "#27AE60"),  # Green
    ("GOV_WGI_CC_EST", "Control of Corruption Index", "#2B4C7E"),  # Blue
    ("NY_GDP_MINR_RT_ZS", "Mineral Rents (% GDP)", "#E67E22"),  # Orange
]

# Set visual theme
sns.set_theme(style="whitegrid")

# ==============================================================================
# GRAPH 1: 2x2 PARTIAL REGRESSION GRID WITH TRIANGLES
# ==============================================================================
fig, axes = plt.subplots(2, 2, figsize=(14, 10), dpi=300)
axes = axes.flatten()

for idx, (var, title, color) in enumerate(sig_vars):
  ax = axes[idx]

  # Plot country-year observations using triangle markers ('^')
  ax.scatter(
      model_data[var],
      model_data[dep_var],
      color=color,
      alpha=0.35,
      marker="^",  # Triangles
      s=30,
      edgecolors="none",
      label="Country-Year Observations",
  )

  # Model partial effect prediction (holding all other predictors at median)
  x_grid = np.linspace(model_data[var].min(), model_data[var].max(), 100)
  pred_df = pd.DataFrame({v: [model_data[v].median()] * 100 for v in indep_vars})
  pred_df[var] = x_grid
  pred_X = sm.add_constant(pred_df, has_constant="add")[X.columns]

  y_pred = model.predict(pred_X)
  ax.plot(
      x_grid,
      y_pred,
      color=color,
      linewidth=2.5,
      label="Master Model Partial Effect",
  )

  ax.set_title(
      title, fontsize=12, fontweight="bold", color="#2C3E50", pad=10
  )
  ax.set_xlabel(title, fontsize=10)
  ax.set_ylabel("Protected Areas (% total land/marine)", fontsize=10)
  ax.legend(loc="upper left", frameon=True, fontsize=9)

plt.suptitle(
    "Master Regression: Statistically Significant Predictors of Protected"
    " Areas",
    fontsize=14,
    fontweight="bold",
    y=0.98,
)
plt.tight_layout(rect=[0, 0, 1, 0.96])
plt.savefig("significant_predictors_grid.png", bbox_inches="tight")
plt.show()

# ==============================================================================
# GRAPH 2: FOREST PLOT (ONLY SIGNIFICANT VARIABLES, TRIANGLES & COLORS)
# ==============================================================================
# Standardize predictors to plot comparative standardized betas
X_std = (
    model_data[indep_vars] - model_data[indep_vars].mean()
) / model_data[indep_vars].std()
X_std = sm.add_constant(X_std)
model_std = sm.OLS(y, X_std).fit(cov_type="HC3")

sig_keys = [v[0] for v in sig_vars]
std_summary = pd.DataFrame({
    "Variable": [v[1] for v in sig_vars],
    "Code": sig_keys,
    "Color": [v[2] for v in sig_vars],
    "Std_Beta": model_std.params[sig_keys].values,
    "CI_Lower": model_std.conf_int().loc[sig_keys, 0].values,
    "CI_Upper": model_std.conf_int().loc[sig_keys, 1].values,
})

fig_f, ax_f = plt.subplots(figsize=(9, 4.5), dpi=300)

for i, row in std_summary.iterrows():
  ax_f.errorbar(
      x=row["Std_Beta"],
      y=i,
      xerr=[
          [row["Std_Beta"] - row["CI_Lower"]],
          [row["CI_Upper"] - row["Std_Beta"]],
      ],
      fmt="^",  # Triangle marker
      color=row["Color"],
      ecolor=row["Color"],
      elinewidth=2.5,
      capsize=5,
      markersize=10,
      label=row["Variable"],
  )

ax_f.axvline(x=0, color="gray", linestyle="--", linewidth=1)
ax_f.set_yticks(range(len(std_summary)))
ax_f.set_yticklabels(std_summary["Variable"], fontsize=11)
ax_f.set_xlabel(
    "Standardized Effect Size (Change in Protected Area % per 1 Std. Dev."
    " Increase)",
    fontsize=10,
)
ax_f.set_title(
    "Master Regression: Statistically Significant Effects (p < 0.001)",
    fontsize=12,
    weight="bold",
    pad=12,
)
ax_f.invert_yaxis()
plt.tight_layout()
plt.savefig("significant_forest_plot_triangles.png", bbox_inches="tight")
plt.show()
```

```python
import pandas as pd
import statsmodels.api as sm
import plotly.express as px

# 1. Load Data
df = pd.read_csv("combined_wdi.csv")

AGGREGATES = {
    'AFE','AFW','ARB','CEB','CSS','EAP','EAR','EAS','ECA','ECS','EMU','EUU',
    'FCV','HIC','HPC','IBD','IBT','IDA','IDB','IDX','LAC','LCN','LDC','LIC',
    'LMC','LMY','LTE','MEA','MIC','MNA','NAC','OED','OSS','PRE','PSS','PST',
    'SAS','SSA','SSF','SST','TEA','TEC','TLA','TMN','TSA','TSS','UMC','WLD'
}

df_c = df[~df['country_code'].isin(AGGREGATES)].copy()

# Fit OLS model on clean panel (2013-2023)
df_panel = df_c[df_c["year"].between(2013, 2023)].copy()
dep_var = "ER_PTD_TOTL_ZS"
indep_vars = [
    "SL_TLF_TOTL_FE_ZS", "SG_GEN_PARL_ZS", "GOV_WGI_PV_SC",
    "GOV_WGI_CC_EST", "NY_GDP_MINR_RT_ZS", "NY_GDP_TOTL_RT_ZS"
]

clean_panel = df_panel[["country_code", "country", "year", dep_var] + indep_vars].dropna()
X = sm.add_constant(clean_panel[indep_vars])
y = clean_panel[dep_var]
model = sm.OLS(y, X).fit(cov_type="HC3")

# 2. Get LATEST AVAILABLE value up to 2023 for each country (fixes blank map issue)
df_latest = (
    df_c[df_c["year"] <= 2023]
    .dropna(subset=[dep_var] + indep_vars)
    .sort_values("year")
    .groupby("country_code")
    .last()
    .reset_index()
)

# Predict values & calculate model accuracy
X_latest = sm.add_constant(df_latest[indep_vars])
df_latest["Predicted_Protection"] = model.predict(X_latest).round(1)
df_latest["Actual_Protection"] = df_latest[dep_var].round(1)
df_latest["Absolute_Error"] = (df_latest["Actual_Protection"] - df_latest["Predicted_Protection"]).abs()
df_latest["Model_Accuracy_%"] = (100 - df_latest["Absolute_Error"]).clip(lower=0, upper=100).round(1)

# 3. Plot Interactive Map with Deep Green Scale
fig = px.choropleth(
    df_latest,
    locations="country_code",
    color="Actual_Protection",
    hover_name="country",
    hover_data={
        "country_code": False,
        "Actual_Protection": ":.1f%",
        "Predicted_Protection": ":.1f%",
        "year": True
    },
    color_continuous_scale="Greens",  # Deeper green for higher percentages
    range_color=[0, df_latest["Actual_Protection"].max()],
    title="Protected Area Coverage (% of Total Territory) & OLS Accuracy",
    labels={
        "Actual_Protection": "Actual Protected %",
        "Predicted_Protection": "Model Predicted %",
        "year": "Data Year"
    }
)

fig.update_layout(
    margin={"r": 0, "t": 40, "l": 0, "b": 0},
    geo=dict(showframe=False, showcoastlines=True, projection_type="natural earth")
)

fig.show()
```

```python
# 1. Load data and run master model
df = pd.read_csv("combined_wdi_countries_only.csv")
df_panel = df[df["year"].between(2013, 2023)].copy()

dep_var = "ER_PTD_TOTL_ZS"
indep_vars = [
    "SL_TLF_TOTL_FE_ZS",  # Female Labor Force Share
    "SG_GEN_PARL_ZS",  # Women in Parliament
    "GOV_WGI_PV_SC",  # Political Stability
    "GOV_WGI_CC_EST",  # Control of Corruption
    "NY_GDP_MINR_RT_ZS",  # Mineral Rents
    "NY_GDP_TOTL_RT_ZS",  # Total Resource Rents
]

model_data = df_panel[["country_code", "country", "year", dep_var] + indep_vars].dropna()

X = sm.add_constant(model_data[indep_vars])
y = model_data[dep_var]
model = sm.OLS(y, X).fit(cov_type="HC3")


# 2. APA Helper Function for p-values (omits leading zero per APA style)
def format_apa_p(p_val):
  if p_val < 0.001:
    return "< .001"
  else:
    return f"= {p_val:.3f}".replace("0.", ".")


# 3. Overall Model Fit Extraction (APA Style)
n_obs = int(model.nobs)
df_model = int(model.df_model)
df_resid = int(model.df_resid)
f_stat = model.fvalue
f_pvalue = model.f_pvalue
r_squared = model.rsquared
adj_r_squared = model.rsquared_adj

# Formatted APA Model Summary Paragraph
apa_model_summary = (
    f"The overall regression model was statistically significant, "
    f"F({df_model}, {df_resid}) = {f_stat:.2f}, p {format_apa_p(f_pvalue)}, "
    f"R² = {r_squared:.3f} (Adjusted R² = {adj_r_squared:.3f}), N = {n_obs:,}."
)

print("=== OVERALL MODEL FIT (APA 7TH) ===")
print(apa_model_summary)
print("\n" + "=" * 50 + "\n")

# 4. Predictor-Level Extraction (APA Style)
apa_rows = []
for var in X.columns:
  b = model.params[var]
  se = model.bse[var]
  t_val = model.tvalues[var]
  p_val = model.pvalues[var]
  ci_lower, ci_upper = model.conf_int().loc[var]

  is_sig = "Significant" if p_val < 0.05 else "Not Significant"

  # APA inline reporting string: b = X.XX, SE = X.XX, t(df) = X.XX, p = .XXX, 95% CI [LB, UB]
  apa_string = (
      f"b = {b:.2f}, SE = {se:.2f}, t({df_resid}) = {t_val:.2f}, "
      f"p {format_apa_p(p_val)}, 95% CI [{ci_lower:.2f}, {ci_upper:.2f}]"
  )

  apa_rows.append({
      "Variable": var,
      "b": f"{b:.4f}",
      "SE": f"{se:.4f}",
      "t": f"{t_val:.2f}",
      "p": format_apa_p(p_val),
      "Significance (α = .05)": is_sig,
      "APA Reported String": apa_string,
  })

apa_df = pd.DataFrame(apa_rows)

print("=== INDIVIDUAL PREDICTORS TABLE (APA 7TH) ===")
print(
    apa_df[
        ["Variable", "b", "SE", "t", "p", "Significance (α = .05)"]
    ].to_string(index=False)
)
```

```python
import statsmodels.api as sm

# 1. Load dataset and filter for panel study window (2013-2023)
df = pd.read_csv("combined_wdi_countries_only.csv")
df_panel = df[df["year"].between(2013, 2023)].copy()

# 2. Define model variables
dep_var = "ER_PTD_TOTL_ZS"
indep_vars = [
    "SL_TLF_TOTL_FE_ZS",  # Female Labor Force Share (%)
    "SG_GEN_PARL_ZS",  # Women in Parliament (%)
    "GOV_WGI_PV_SC",  # Political Stability Score
    "GOV_WGI_CC_EST",  # Control of Corruption Index
    "NY_GDP_MINR_RT_ZS",  # Mineral Rents (% GDP)
    "NY_GDP_TOTL_RT_ZS",  # Natural Resource Rents (% GDP)
]

# 3. Clean complete cases for model estimation
model_data = df_panel[
    ["country_code", "country", "year", dep_var] + indep_vars
].dropna()

# 4. Fit the Master OLS Model (HC3 Robust)
X = sm.add_constant(model_data[indep_vars])
y = model_data[dep_var]
model = sm.OLS(y, X).fit(cov_type="HC3")

# 5. Calculate model predictions and the Protection Gap (Residual = Actual - Predicted)
model_data["Predicted_Protection"] = model.predict(X)
model_data["Protection_Gap"] = (
    model_data[dep_var] - model_data["Predicted_Protection"]
)

# 6. Aggregate across the panel window by country
country_summary = (
    model_data.groupby(["country_code", "country"])
    .agg(
        Actual_Protection=(dep_var, "mean"),
        Predicted_Protection=("Predicted_Protection", "mean"),
        Protection_Gap=("Protection_Gap", "mean"),
        Control_of_Corruption=("GOV_WGI_CC_EST", "mean"),
        Women_in_Parliament=("SG_GEN_PARL_ZS", "mean"),
        Female_Labor_Force=("SL_TLF_TOTL_FE_ZS", "mean"),
    )
    .reset_index()
)

# 7. Extract top 3 countries with the largest negative Protection Gap
top3_countries = country_summary.sort_values("Protection_Gap").head(3).copy()


from IPython.display import display

# Format the extracted top 3 candidates into a styled DataFrame
top3_formatted = pd.DataFrame({
    "Country": top3_countries["country"],
    "Actual Protection (%)": top3_countries["Actual_Protection"].map(
        "{:.2f}%".format
    ),
    "Model Predicted Protection (%)": top3_countries[
        "Predicted_Protection"
    ].map("{:.2f}%".format),
    "Protection Gap (Residual)": top3_countries["Protection_Gap"].map(
        "{:.2f}%".format
    ),
    "Control of Corruption Score": top3_countries["Control_of_Corruption"].map(
        "{:.2f}".format
    ),
    "Women in Parliament (%)": top3_countries["Women_in_Parliament"].map(
        "{:.1f}%".format
    ),
})

# Display as an interactive Jupyter Notebook table
display(top3_formatted)
```

An important point to note about the data is that the statistic does not include the unpaid labour of women raising kids, cleaning the house and familial administration, hence there are women working who are not included.

While our study offers key empirical insights into the drivers of territorial protection, several limitations must be acknowledged. 

<strong>First</strong>, our model relies on a concise set of seven World Bank indicators, not a lot of indicators in the grand scheme of things. Furthermore, our dataset may fail to capture informal, indigenous, or community-managed conservation areas due to strict reporting protocols. This creates notable data gaps for nations like Cabo Verde, where active coastal protection is not fully reflected in official percentage metrics. 

<strong>Second</strong>, our natural resource metrics focus on gross rents as a percentage of GDP, which omits economic factors such as sector-specific employment, value added to the domestic economy, and competing industries like agricultural or marine.

<strong>Third</strong>, percentage-based targets like the 30x30 goal treat all landmasses as equal, masking vast physical differences where protecting 30% of a land-abundant nation like the United States represents millions of square kilometers, but 100% of a microstate like Luxembourg yields much less.

<h3 style="margin-top: 0;">Explaining our Model</h3>

<p>
We were curious to explore our data further as we were surprised by some of the results from our linear regression model. We were surprised that the Naural Resource Rents alongside Political Stability were not stitistically signifacnt indicators, so we decided to explore the relationship between variables further.
    
Unlike linear regression which only measures the relationship between two variables, our random forest regressor enables us to measure the non-linear relationships and interactions between our variables.

Our model builds a forest of 100 decision trees.

It is then able to look at random branches of our country-year cases, splitting it into different paths.

So <strong>in order to predict a country's protected area percentage, our forest regressor passes the country's indicators through all the different decision trees and averages final predictions<strong>.
</p>

```python
#Machine learning
```

```python
import numpy as np
import pandas as pd
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_squared_error, r2_score
from sklearn.model_selection import train_test_split

# 1. Load data and filter panel (2013-2023)
df = pd.read_csv("combined_wdi_countries_only.csv")
df_panel = df[df["year"].between(2013, 2023)].copy()

dep_var = "ER_PTD_TOTL_ZS"  # Target variable
indep_vars = [
    "SL_TLF_TOTL_FE_ZS",  # Female Labor Force Share
    "SG_GEN_PARL_ZS",  # Women in Parliament
    "GOV_WGI_PV_SC",  # Political Stability
    "GOV_WGI_CC_EST",  # Control of Corruption
    "NY_GDP_MINR_RT_ZS",  # Mineral Rents
    "NY_GDP_TOTL_RT_ZS",  # Total Resource Rents
]

# 2. Clean missing values and isolate X and y
model_data = df_panel[[dep_var] + indep_vars].dropna()
X = model_data[indep_vars]
y = model_data[dep_var]

# 3. Train-Test Split (80% Train, 20% Test)
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42
)

# 4. Initialize and fit the Random Forest Model
model = RandomForestRegressor(n_estimators=100, random_state=42)
model.fit(X_train, y_train)

# 5. Print evaluation metrics
y_pred = model.predict(X_test)
r2 = r2_score(y_test, y_pred)
rmse = np.sqrt(mean_squared_error(y_test, y_pred))
print(f"Training set size: {len(X_train)} | Test set size: {len(X_test)}")
print(f"Test R² Accuracy:  {r2:.2%}")
print(f"Test RMSE Error:   {rmse:.2f}%")
```

This project investigated how gender equality, governance, and natural resource dependence relate to protected-area coverage across countries between 2013 and 2023.
